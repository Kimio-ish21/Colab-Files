Recursion notes
Recursion is a way of solving a problem where a function calls itself inside a function

```python
#Create a funcion theat prints the numbers 1 through n

def printNums(n):
  for i in range(1, n+1):
    print(i)
#Recursivley
def printNums_r(n):
  #first identify base casee
  if n == 0:
    return
  else:
    printNums_r(n-1)
    print(n)
print(printNums_r(6))

#Write a function to find factorial
def factorial(n):
  if n == 1:
      return 1
  else:
    return n * factorial(n - 1)

print(factorial(6))

#Works:
#F_r(5) = 5* f_4 120
#F_r(4) = 4* f3 24
#F_r(3) = 3* f_2 6
#F_r(2) = 2* f_1 2
#F_r(1) = --> 1
```

#List Comprehesnion

- A shorter way to create lists
- Instead of writing loops and appending, you create a new list in one line

```python
#Old way
numbers = [1, 2, 3, 4, 5, 6]
evenList = []
for i in numbers:
  if i %2 == 0:
    evenList.append(i)
print(evenList)
```

```python
#New way
evenListLC = [x for x in numbers if x%2 ==0]
print(evenListLC)
squaredLC = [x**2 for x in numbers]
print(squaredLC)
```

```python
import string

# 1. Start with input
message = input("Enter your message: ")
shift = int(input("Enter shift value: "))

# 2. Shift mod 26
shift = shift % 26

# 3. Create alphabet index and dictionary for translation map
alphabet = string.ascii_lowercase
shift_dict = {}

# Build the map: {original_char: shifted_char}
for i in range(len(alphabet)):
    # Calculate shifted index with wrap-around
    shifted_char = alphabet[(i + shift) % 26]
    shift_dict[alphabet[i]] = shifted_char

# 4. Process each character in the user message
encrypted_message = ""

for char in message:
    if char.isalpha():
        # Handle lowercase character check
        is_upper = char.isupper()
        lower_char = char.lower()

        # Set the encrypted character from the dictionary
        new_char = shift_dict[lower_char]

        # Restore case if original was uppercase
        encrypted_message += new_char.upper() if is_upper else new_char
    else:
        # Non-alphabetic characters stay unchanged
        encrypted_message += char

print(f"Encrypted Message: {encrypted_message}")
```

Error Handling

If you don't handle errors, your program stops running. It gives you feedback to prevent crashes.
Try Except Handles errors
try:
  #code
except:
  #code that runs if error happens

```python
try:
  num1 = int(input("Give me a number\n"))
  print(f'Oh {num1} is my favorite number')
except ValueError:
  print("That is not a number")
```

Most common Error Types:
1. ValueError - Wrong types of input
2. ZeroDivisionError - Can't divide by 0
3. IndexError - List of indexout of range
4. Generic - Create your own

```python
print("I'll divide 10 to the number you give")
while True:
  try:
    div = int(input("Give me a number\n"))
    ans = 10/div
    print(ans)
    break
  except ZeroDivisionError:
    print("Can't divide it by that")
```
